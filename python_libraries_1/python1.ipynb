```python
import pandas as pd
import numpy as np

df = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv", sep='\t')
```

```python
# Display the last 7 rows of the DataFrame
df.head(7)
```

```python
# Print a summary of the df including the index dtype and columns dtypes, non-null values, and memory usage.

df.info()
```

```python
# Determine how many observations have missing item_price in the dataset

int(df['item_price'].isnull().sum())
```

```python
# Print the names of the columns in the df

list(df.columns)
```

```python
# Identify which item_name appears most frequently in the orders

df.groupby('item_name').size().sort_values(ascending=False).head(1)
```

```python
# Determine how many distinct item_name values are present in the dataset
list(df['item_name'].unique())
```

```python
# Create a new df containing only the rows where the item_name is Chicken Bowl

chicken_bowl_df = df[df['item_name'] == 'Chicken Bowl']
chicken_bowl_df.head(5)
```

```python
# Convert item_price from string to float and remove the dollar sign
df['item_price'] = df['item_price'].str.replace('$', '').astype(float)
df.head(5)
```

```python
# Calculate the avg item_price across all orders

avg_item_price = df['item_price'].mean()
print(f"The average item price across all orders is: ${avg_item_price:.2f}")
```

```python
# Determine the total number of unique orders in the dataset based on the order_id column

unique_orders = df['order_id'].unique()
print(f"The total number of unique orders in the dataset is: {len(unique_orders)}")
```
