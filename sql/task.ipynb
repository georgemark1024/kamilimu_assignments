```python
import sqlite3
import pandas as pd

# Load datasets
athletes = pd.read_csv("https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-07-27/athlete_events.csv")
regions = pd.read_csv("https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-07-27/noc_regions.csv")
sales = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv")
sales = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv", sep='\t')

# Connect to SQLite in-memory DB
conn = sqlite3.connect(":memory:")

# Write DataFrames to SQL tables
athletes.to_sql("athletes_table", conn, index=False, if_exists="replace")
regions.to_sql("regions_table", conn, index=False, if_exists="replace")
sales.to_sql("sales_table", conn, index=False, if_exists="replace")
```

```python
 # Total number of medals won by each country showing top 5
print("Athletes Table:")
pd.read_sql(
    """SELECT count(Medal) as total_medals, NOC
    FROM athletes_table
    GROUP BY NOC
    ORDER BY total_medals DESC
    LIMIT 5
    """, conn)
# print("Regions Table:")
# pd.read_sql("SELECT * FROM regions_table LIMIT 5", conn)
# print("Sales Table:")
# pd.read_sql("SELECT * FROM sales_table LIMIT 5", conn)
```

```python
# Average age of atheletes who won GOLD medals
print("Average age of athletes who won GOLD medals:")
pd.read_sql(
    """SELECT AVG(Age) as average_age
    FROM athletes_table
    WHERE Medal = 'Gold'
    """, conn)
```

```python
# How many distinct events are there in each sport
print("Number of distinct events in each sport:")
pd.read_sql(
    """SELECT Sport, COUNT(DISTINCT Event) as distinct_events
    FROM athletes_table
    GROUP BY Sport
    """, conn)
```

```python
# All atheletes from NOC="USA"
print("All athletes from NOC='USA':")
pd.read_sql(
    """SELECT * 
    FROM athletes_table
    WHERE NOC = 'USA'
    """, conn
)
```

```python
# Count how many medals were awarded each year
pd.read_sql(
    """SELECT Year, COUNT(Medal) AS total_medals
    FROM athletes_table
    GROUP BY Year
    """, conn
)
```

```python
# All missing height or weight records for athletes
pd.read_sql(
    """SELECT *
    FROM athletes_table
    WHERE Height IS NULL OR Weight IS NULL
""", conn
)
```

```python
# Replace missing height with the average athlete height
athletes_copy = athletes.copy()
avg_height = athletes_copy['Height'].mean()

print(f"Before fillna")
null = athletes['Height'].isna().to_list()
print(f"Null values = {null.count(True)}")

athletes_copy['Height'].fillna(value=avg_height)

print("After fillna")
null = athletes['Height'].isna().to_list()
print(f"Null values = {null.count(True)}")

print(athletes_copy)

#################################################################
"""To be Done"""
#################################################################
```
